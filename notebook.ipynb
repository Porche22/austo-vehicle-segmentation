### Univariate Analysis

#### **Question 6:** Explore all the variables and provide observations on their distributions. (Generally, histograms, boxplots, countplots, etc. are used for univariate exploration.) [10 marks]

```python
# Function to plot histograms and boxplots for numerical variables
def univariate_num(data, col, title):
    fig, ax = plt.subplots(1, 2, figsize=(14, 5))

    # Histogram
    sns.histplot(data=data, x=col, kde=True, ax=ax[0], color='teal')
    ax[0].set_title(f'Distribution of {title}')

    # Boxplot
    sns.boxplot(data=data, x=col, ax=ax[1], color='orange')
    ax[1].set_title(f'Boxplot of {title}')

    plt.tight_layout()
    plt.show()

# Function to plot countplots for categorical variables
def univariate_cat(data, col, title):
    plt.figure(figsize=(8, 5))
    ax = sns.countplot(data=data, x=col, palette='viridis', order=data[col].value_counts().index)
    plt.title(f'Count of {title}')

    # Add percentage labels
    total = len(data)
    for p in ax.patches:
        percentage = f'{100 * p.get_height() / total:.1f}%'
        x = p.get_x() + p.get_width() / 2 - 0.05
        y = p.get_height()
        ax.annotate(percentage, (x, y), ha='center', va='bottom')
    plt.show()
```

**Numerical Variables Analysis**

```python
# Analyze Age
univariate_num(df, 'Age', 'Age')


# Analyze Salaries
univariate_num(df, 'Salary', 'Customer Salary')
univariate_num(df, 'Total_salary', 'Total Household Income')


# Analyze Price
univariate_num(df, 'Price', 'Car Price')
```

#### Question 7: How many cars are of make Hatchback and priced above 25000. State your observations? [2 marks]

```python
print("\n[Q8] Owners with Car Price > Salary & Personal Loan status")

# 1. Owners with Price > Salary
price_gt_salary = df[df['Price'] > df['Salary']]
count_high_spend = price_gt_salary.shape[0]

# 2. Of these, how many took a Personal Loan?
loan_takers = price_gt_salary[price_gt_salary['Personal_loan'] == 'Yes'].shape[0]

print(f"Number of owners (Price > Salary): {count_high_spend}")
print(f"Of these, number with Personal Loan: {loan_takers}")
if count_high_spend > 0:
    print(f"Percentage taking loan: {(loan_takers/count_high_spend)*100:.2f}%")
```

#### Question 10:For customers who have 3 or fewer dependents, how does the average car price vary by profession ? [2 marks]

```python
print("\n[Q10] For customers with 3 or fewer dependents, how does avg price vary by profession?")

# Filter: 3 or fewer dependents
# FIX: Use 'No_of_Dependents' (capital 'D') to match dataset column name
df_small_fam = df[df['No_of_Dependents'] <= 3]

# Group by Profession and calculate mean Price
avg_price_prof = df_small_fam.groupby('Profession')['Price'].mean().sort_values()

print("\nAverage Price by Profession (Dependents <= 3):")
print(avg_price_prof)

# Visualization
plt.figure(figsize=(8, 5))
sns.barplot(x=avg_price_prof.index, y=avg_price_prof.values, palette='magma')
plt.title('Avg Car Price by Profession (Dependents <= 3)')
plt.ylabel('Average Price')
plt.show()
```

**Observations:**

  1. **Salaried Spend More:** Contrary to common assumptions, Salaried professionals with small families spend significantly more Avg 37,036 compared to Business professionals 32,830.

  2. **Strategic Pivot:** The Salaried segment is not just the "Volume" market; they are also the "Value" market for mid-tier vehicles. They likely prioritize cars as lifestyle upgrades more than Business owners in this demographic.

  3. **Targeting:** Focus "Premium Upgrade" marketing campaigns on Salaried professionals, as they show a higher willingness to pay.

#### Question 11: For customers who have availed a home loan and a personal loan, how does the price vary by profession? [3 marks]

```python
print("\n[Q11] Price variation by Profession for customers with BOTH loans")

# Filter: Personal_loan = Yes AND House_loan = Yes
double_loans = df[(df['Personal_loan'] == 'Yes') & (df['House_loan'] == 'Yes')]

if not double_loans.empty:
    avg_price_double = double_loans.groupby('Profession')['Price'].mean()
    print("\nAverage Price (Double Loans):")
    print(avg_price_double)

    plt.figure(figsize=(8, 5))
    sns.barplot(x=avg_price_double.index, y=avg_price_double.values, palette='rocket')
    plt.title('Avg Car Price by Profession (Double Loans)')
    plt.show()
else:
    print("No customers with both loans found.")
```

## Conclusion and Recommendations

#### **Question 12:** Write the conclusions and business recommendations derived from the analysis. (6 marks)

#### **A. Market Segmentation & Buyer Profiles**

  1. **The "Premium Hatchback" Opportunity (Based on Q7):**

     * **Insight:** **50.23%** of Hatchback sales are >$25k. This is a massive, premium sub-segment.
     * **Strategy:** Do not market Hatchbacks as "Cheap." Launch a specific "Sport/Lux" Hatchback trim to cater to this 50%. Feature-led marketing (Sunroof, Tech) will work better than Price-led marketing here.
     * **Action Item:** Create two distinct brochure types for Hatchbacks: "Efficiency" (for the <$25k segment) and "Performance" (for the >$25k segment).
  2. **The High-Value Salaried Professional (Based on Q10 & Q11):**

     * **Insight:** Across key segments (Small Family, High Debt), **Salaried professionals consistently outspend Business owners** (by ~$2k - $4k).
     * **Profile:** Likely corporate employees who view cars as status symbols and are comfortable with monthly payments (EMI).
     * **Strategy:** They are the primary target for **Up-selling**. Pitch them the "Top Model" Sedan or SUV. They are less price-sensitive than Business owners in comparable situations.
  3. **The Practical Business Buyer (Based on Q10):**

     * **Insight:** Business owners spending less (~$32k) suggests a focus on utility or cost-consciousness in this dataset.
     * **Strategy:** Pitch "Value," "Low Maintenance," and "Durability" to the Business segment. Highlight "Total Cost of Ownership" rather than monthly EMI.



#### **B. Financing Strategy**

  1. **Leveraging Debt Comfort (Based on Q8 & Q11):**
     * **Insight:** Salaried employees maintain high spending even with loans. 46% of high-spenders take loans.
     * **Action:** Launch **"Consolidated EMI"** plans. Since they likely have other loans, offer a car loan structure that starts low and increases (Step-up EMI) to fit their cash flow.



#### **C. Portfolio & Customer Lifecycle**

  1. **Inventory Planning:**

     * **Stock Depth:** Ensure high availability of **Mid-to-High trim Sedans** for the Salaried class.
     * **SUVs:** Keep as the niche flagship (only ~20% of market), targeting the ultra-high income bracket (Total Salary >$100k) identified in Q9.
  2. **Lifecycle Management (The "Austo Upgrade" Path):**

     * **Insight:** The clear age progression (Hatchback -> Sedan -> SUV) is a retention goldmine.
     * **Action:** Implement a **3-Year Upgrade Program**. Target Hatchback buyers (Median age 28) with guaranteed buyback offers when they hit age 31-32 (the typical Sedan entry age), making it seamless for them to trade in their Hatchback for a new Austo Sedan.



#### **D. Marketing Channels**

  1. **Channel Mix:**
     * **Digital (Instagram/TikTok):** Focus exclusively on the Entry-Level Hatchback ($15k-$25k) targeting the <30 demographic.
     * **LinkedIn:** The primary channel for the **Premium Salaried** segment ($35k+). Advertise "Executive Sedans" and "Status" here.
     * **Local Business Forums:** Target the Business segment with "Fleet/Utility" offers focusing on depreciation benefits.
